```python
import pandas as pd
```

```python
df = pd.read_csv("../01_Raw_Data/customer_shopping_behavior_original.csv")
```

```python
bins = [0, 19, 39, 59, 79, 100]
labels = ['Under 20', '20-39', '40-59', '60-79', '80-100']

df['Age Group'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
frequency_score_map = {
    'Annually': 1,
    'Every 3 Months': 1,
    'Quarterly': 2,
    'Monthly': 3,
    'Fortnightly': 3,
    'Bi-Weekly': 4,
    'Weekly': 4
}

df['Frequency Score'] = df['Frequency of Purchases'].map(frequency_score_map)
```

```python
bins = [0, 39, 59, 79, 100]
labels = [1, 2, 3, 4]

df['Purchase Value Score'] = pd.cut(
    df['Purchase Amount (USD)'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
df['Customer Behavior Score'] = (
    df['Frequency Score'].astype(int)
    + df['Purchase Value Score'].astype(int)
)
```

```python
def assign_segment(score):
    if score <= 3:
        return 'Low Engagement'
    elif score <= 5:
        return 'Occasional'
    elif score <= 7:
        return 'Regular'
    else:
        return 'High Value'

df['Customer Segment'] = df['Customer Behavior Score'].apply(assign_segment)
```

```python
df['Customer Segment'].value_counts()
```

```python
import matplotlib.pyplot as plt
```

```python
segment_counts = df['Customer Segment'].value_counts()

segment_counts.plot(kind='bar', figsize=(8,5))

plt.title('Customer Segment Distribution')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
segment_revenue = (
    df.groupby('Customer Segment')['Purchase Amount (USD)']
      .sum()
      .sort_values(ascending=False)
)

segment_revenue.plot(kind='bar', figsize=(8,5))

plt.title('Total Revenue by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Purchase Amount (USD)')
plt.xticks(rotation=0)
plt.show()
```

```python
segment_avg = (
    df.groupby('Customer Segment')['Purchase Amount (USD)']
      .mean()
      .sort_values(ascending=False)
)

segment_avg.plot(kind='bar', figsize=(8,5))

plt.title('Average Purchase Amount by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Average Purchase Amount (USD)')
plt.xticks(rotation=0)
plt.show()
```

```python
category_revenue = (
    df.groupby('Category')['Purchase Amount (USD)']
      .sum()
      .sort_values(ascending=False)
)

category_revenue.plot(kind='bar', figsize=(8,5))

plt.title('Total Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Purchase Amount (USD)')
plt.xticks(rotation=0)
plt.show()
```

```python
age_revenue = (
    df.groupby('Age Group', observed=True)['Purchase Amount (USD)']
      .sum()
      .sort_values(ascending=False)
)

age_revenue.plot(kind='bar', figsize=(8,5))

plt.title('Total Revenue by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Total Purchase Amount (USD)')
plt.xticks(rotation=0)
plt.show()
```

```python
frequency_avg = (
    df.groupby('Frequency of Purchases')['Purchase Amount (USD)']
      .mean()
      .sort_values(ascending=False)
)

frequency_avg.plot(kind='bar', figsize=(9,5))

plt.title('Average Purchase Amount by Purchase Frequency')
plt.xlabel('Purchase Frequency')
plt.ylabel('Average Purchase Amount (USD)')
plt.xticks(rotation=45, ha='right')
plt.show()
```
